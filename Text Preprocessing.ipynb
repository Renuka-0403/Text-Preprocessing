```python
# ============================================================
# TEXT PREPROCESSING USING PYTHON
# PART 1 : IMPORT LIBRARIES & LOAD DATASET
# ============================================================

import pandas as pd
import re
import string
import nltk

from nltk.tokenize import sent_tokenize, word_tokenize

print("=" * 60)
print("TEXT PREPROCESSING PROJECT")
print("=" * 60)

# ------------------------------------------------------------
# Read the CSV Dataset
# ------------------------------------------------------------

df = pd.read_csv("Raw_Text.csv", encoding="utf-8")

print("\n")
print("=" * 60)
print("ORIGINAL DATASET")
print("=" * 60)

print(df)

print("\nTotal Records :", len(df))
print("Total Columns :", len(df.columns))

# ------------------------------------------------------------
# Create Empty List to Store Results
# ------------------------------------------------------------

processed_data = []

print("\nDataset Loaded Successfully.")
# ============================================================
# PART 2 : LOWERCASE, SENTENCE TOKENIZATION & WORD TOKENIZATION
# ============================================================

print("\n")
print("=" * 60)
print("TEXT PREPROCESSING")
print("=" * 60)

for index, row in df.iterrows():

    text = str(row["Text"])

    # Convert to Lowercase
    lower_text = text.lower()


    try:
        sentences = sent_tokenize(lower_text)
    except:
        sentences = re.split(r'[.!?]+', lower_text)
        sentences = [s.strip() for s in sentences if s.strip()]

    words = []
    for sentence in sentences:
        try:
            tokens = word_tokenize(sentence)
        except:
            tokens = re.findall(r"\b[\w'-]+\b", sentence)

        words.extend(tokens)

    processed_data.append({
        "ID": row["ID"],
        "Original_Text": text,
        "Lowercase_Text": lower_text,
        "Sentence_Tokens": sentences,
        "Word_Tokens": words
    })

print("Lowercase Conversion Completed.")
print("Sentence Tokenization Completed.")
print("Word Tokenization Completed.")

print("\n")
print("=" * 60)
print("SAMPLE OUTPUT")
print("=" * 60)
sample = processed_data[0]

print("ID :", sample["ID"])
print("\nOriginal Text:")
print(sample["Original_Text"])

print("\nLowercase Text:")
print(sample["Lowercase_Text"])

print("\nSentence Tokens:")
print(sample["Sentence_Tokens"])

print("\nWord Tokens:")
print(sample["Word_Tokens"])
print("\n")
print("=" * 60)
print("TEXT CLEANING")
print("=" * 60)

for item in processed_data:
    cleaned_words = []
    for word in item["Word_Tokens"]:
        word = re.sub(r'https?://\S+|www\.\S+', '', word)
        word = re.sub(r'[^\x00-\x7F]+', '', word)
        word = re.sub(r'\d+', '', word)
        word = word.translate(str.maketrans('', '', string.punctuation))
        word = word.strip()
        if word != "":
            cleaned_words.append(word)
    item["Cleaned_Words"] = cleaned_words

print("URLs Removed.")
print("Emojis Removed.")
print("Numbers Removed.")
print("Punctuation Removed.")
print("Extra Spaces Removed.")

print("\n")
print("=" * 60)
print("SAMPLE OUTPUT AFTER CLEANING")
print("=" * 60)
sample = processed_data[0]

print("ID :", sample["ID"])
print("\nOriginal Tokens:")
print(sample["Word_Tokens"])

print("\nCleaned Tokens:")
print(sample["Cleaned_Words"])

print("\n")
print("=" * 60)
print("STOPWORD REMOVAL")
print("=" * 60)

stop_words = {
    "a","an","the","is","am","are","was","were","be","been","being",
    "have","has","had","do","does","did",
    "and","or","but","if","because","as","until","while",
    "of","at","by","for","with","about","against","between","into",
    "through","during","before","after","above","below",
    "to","from","up","down","in","out","on","off","over","under",
    "again","further","then","once",
    "here","there","when","where","why","how",
    "all","any","both","each","few","more","most","other","some","such",
    "no","nor","not","only","own","same","so","than","too","very",
    "can","will","just","should","now",
    "your","you","yours","our","ours","their","they","them",
    "this","that","these","those",
    "it","its","he","him","his","she","her","hers",
    "we","us","i","me","my","mine"
}

for item in processed_data:
    filtered_words = []
    for word in item["Cleaned_Words"]:
        if word.lower() not in stop_words:
            filtered_words.append(word)
    item["Filtered_Words"] = filtered_words

print("Stopwords Removed Successfully.")
print("\n")
print("=" * 60)
print("SAMPLE OUTPUT")
print("=" * 60)

sample = processed_data[0]
print("ID :", sample["ID"])
print("\nCleaned Words:")
print(sample["Cleaned_Words"])

print("\nAfter Stopword Removal:")
print(sample["Filtered_Words"])

print("\n")
print("=" * 60)
print("FINAL CLEANED DATASET")
print("=" * 60)
cleaned_texts = []
for item in processed_data:
    cleaned_sentence = " ".join(item["Filtered_Words"])
    cleaned_texts.append(cleaned_sentence)
cleaned_df = pd.DataFrame({
    "ID": df["ID"],
    "Cleaned_Text": cleaned_texts
})

print(cleaned_df)
print("\n")
print("=" * 60)
print("TOTAL RECORDS")
print("=" * 60)

print("Total Records :", len(cleaned_df))
print("Total Columns :", len(cleaned_df.columns))

cleaned_df.to_csv(
    "Cleaned_Text.csv",
    index=False,
    encoding="utf-8-sig"
)

print("\n")
print("=" * 60)
print("OUTPUT SAVED")
print("=" * 60)

print("File Name : Cleaned_Text.csv")
print("Status    : Saved Successfully")

print("\n")
print("=" * 60)
print("TEXT PREPROCESSING COMPLETED SUCCESSFULLY")
print("=" * 60)
 




```
