```python
import pandas as pd
import re
import string
import nltk

from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer

df = pd.read_csv("Raw Text.txt")
text = " ".join(df["Text"].astype(str))
print(text)

print("="*60)
print("ORIGINAL TEXT")
print("="*60)

print(text)
print("\n")
print("="*60)
print("LOWERCASE CONVERSION")
print("="*60)

lower_text = text.lower()
print(lower_text)

print("\n")
print("="*60)
print("SENTENCE TOKENIZATION")
print("="*60)

try:
    sentences = sent_tokenize(lower_text)
    for i, sentence in enumerate(sentences, 1):
        print(f"{i}. {sentence}")

except:
    sentences = re.split(r'(?<=[.!?])\s+', lower_text)
    for i, sentence in enumerate(sentences, 1):
        print(f"{i}. {sentence}")

print("\n")
print("="*60)
print("WORD TOKENIZATION")
print("="*60)

try:
    words = word_tokenize(lower_text)

except:
    words = re.findall(r"\b[\w'-]+\b", lower_text)

print(words)

print("\n")
print("Total Sentences :", len(sentences))
print("Total Words :", len(words))
print("\n")
print("="*60)
print("REMOVE URLS")
print("="*60)

url_removed = re.sub(r'https?://\S+|www\.\S+', '', lower_text)
print(url_removed)

print("\n")
print("="*60)
print("REMOVE EMAIL IDS")
print("="*60)

email_removed = re.sub(r'\S+@\S+', '', url_removed)
print(email_removed)

print("\n")
print("="*60)
print("REMOVE EMOJIS")
print("="*60)

emoji_removed = re.sub(r'[^\x00-\x7F]+', '', email_removed)
print(emoji_removed)

print("\n")
print("="*60)
print("REMOVE NUMBERS")
print("="*60)

number_removed = re.sub(r'\d+', '', emoji_removed)
print(number_removed)

print("\n")
print("="*60)
print("REMOVE PUNCTUATION")
print("="*60)

punctuation_removed = number_removed.translate(
    str.maketrans('', '', string.punctuation)
)

print(punctuation_removed)
print("\n")
print("="*60)
print("REMOVE EXTRA SPACES")
print("="*60)

clean_text = re.sub(r'\s+', ' ', punctuation_removed).strip()

print(clean_text)
print("\n")
print("="*60)
print("WORD TOKENIZATION AFTER CLEANING")
print("="*60)

tokens = clean_text.split()
print(tokens)
print("\n")
print("="*60)
print("REMOVE STOP WORDS")
print("="*60)

try:
    stop_words = set(stopwords.words('english'))
except:
    stop_words = {
        'a','an','the','is','am','are','was','were','be','been','being',
        'i','me','my','myself','we','our','ours','you','your','he','she',
        'it','they','them','this','that','these','those','to','of','in',
        'on','at','for','from','by','with','and','or','but','as','if',
        'because','while','about','against','between','into','through',
        'during','before','after','above','below','up','down','out','off',
        'over','under','again','further','then','once','here','there',
        'when','where','why','how','all','any','both','each','few','more',
        'most','other','some','such','no','nor','not','only','own','same',
        'so','than','too','very','can','will','just','should','now','your'
    }

filtered_words = [word for word in tokens if word not in stop_words]
print(filtered_words)
print("\n")
print("="*60)
print("STEMMING")
print("="*60)

stemmer = PorterStemmer()
stemmed_words = [stemmer.stem(word) for word in filtered_words]

print(stemmed_words)
print("\n")
print("="*60)
print("LEMMATIZATION")
print("="*60)

lemmatizer = WordNetLemmatizer()

try:
    lemmatized_words = [lemmatizer.lemmatize(word) for word in filtered_words]
except:
    lemmatized_words = filtered_words

print(lemmatized_words)
print("\n")
print("="*60)
print("FINAL CLEAN TEXT")
print("="*60)

final_text = " ".join(lemmatized_words)

print(final_text)

print("\n")
print("="*60)
print("TOTAL WORDS")
print("="*60)

print("Original Words :", len(words))
print("After Cleaning :", len(filtered_words))

with open("Cleaned_Text.txt", "w", encoding="utf-8") as file:
    for sentence in sentences:
        sentence = sentence.lower()
        sentence = re.sub(r'https?://\S+|www\.\S+', '', sentence)
        sentence = re.sub(r'\S+@\S+', '', sentence)
        sentence = re.sub(r'[^\x00-\x7F]+', '', sentence)
        sentence = re.sub(r'\d+', '', sentence)
        sentence = sentence.translate(str.maketrans('', '', string.punctuation))
        sentence = re.sub(r'\s+', ' ', sentence).strip()

        file.write(sentence + "\n")
```
